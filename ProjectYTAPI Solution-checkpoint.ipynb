**Q2b. Provide the following statistics for query string “avatar movie” of top 50 videos sorted by relevance in the US region (regionCode)**

```python
request = youtube.search().list(part="snippet",maxResults=50,q="avatar movie",regionCode="US",type='video')
response = request.execute()
```

```python
videoid_list = []     # Retreiving video ids for all the top 50 videos

for item in response['items']:
    videoid_list.append(item['id']['videoId'])
```

```python
len(videoid_list)   # Displaying number of video ids
```

```python
videoid_list       # Displaying all the video ids
```

```python
request_video = youtube.videos().list( id=','.join(videoid_list),part='snippet,statistics') # Extracting all the videos using video id 
response_video = request_video.execute()
```

```python
pprint(response_video) # Displaying all the video content details
```

```python
video_content = []                     # Creating a list of dictionary to contain all the contents of videos
for video in response_video['items']:
    data = dict(Video_id = video['id'],
                Title = video['snippet']['title'],
                No_of_views = int(video['statistics']['viewCount']),
                No_of_comments = int(video['statistics']['commentCount']),
                No_of_likes = int(video['statistics']['likeCount']),
                )
    video_content.append(data)
```

```python
pprint(video_content)                # Displaying the contents of video
```

```python
df = pd.DataFrame(video_content)  # Creating a Data frame for the video content
```

```python
df
```

```python
df.to_csv('Assigment_1_solution.csv')  # Exporting the Data frame to CSV
```

**Q3d.  Write a function to get the likes vs views ratio of the top 10 videos obtained in 3.a with the highest comments**

```python
def like_vs_view_ratio():        # Function to create a list of ratio for top 10 videos
    top10_ratio = []
    for videos in top10_videos:
        likes = videos['No_of_likes']
        views = videos['No_of_views']
        ratio = likes / views if views > 0 else 0.0
        data = dict(Video_Id = videos['Video_id'],
                    Video_title = videos['Title'],
                    Ratio = ratio)
        top10_ratio.append(data)
    return top10_ratio
    
```

```python
top10_ratio = like_vs_view_ratio() # Calling the function 
pprint(top10_ratio)
```
